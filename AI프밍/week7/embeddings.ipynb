```python
import torch
import numpy as np
import math
```

Embedding for Tokens

```python
# Embedding Paramters
vocab_size = 20
d_model = 6
shape = (vocab_size, d_model)
ep = torch.randn(shape, dtype=torch.float32, requires_grad=True)
ep.shape
```

```python
# input ids
batch_size = 4
seq_len = 5
input_ids = torch.randint(0, vocab_size, (batch_size, seq_len))
input_ids
```

```python
# get embedding vectors for each token in input_ids
bsd = ep[input_ids]# 고오오급 인덱싱! 인덱싱파트랑 복습같이 해보기!
bsd.shape
```

Positional Embedding

```python
max_len = seq_len#5
pe = torch.zeros(max_len, d_model, dtype=torch.float32)#벡터 0으로 초기화해서 생성 [5,6]
print(f"position embedding shape = {pe.shape}")

position = torch.arange(0, max_len, dtype=torch.float32).unsqueeze(1) #shape([5,1])[[0.],[1.],[2.],[3.],[4.]]
div_term = torch.exp(torch.arange(0, d_model, 2).float() / d_model * (-math.log(10000)))
print(f"div shape:{div_term}")
pe[ : , 0: :2] = torch.sin(position * div_term)#슬라이싱을 통한 sin cos 홀짝홀짝 삽입
pe[:, 1::2] = torch.cos(position * div_term)

pe = pe.unsqueeze(dim=0)
print(f"augmented position embedded shape = {pe.shape}")
print(pe)
```

```python
# get position embedding vector

batch_size = 2
seq_len = 5
input_ids = torch.randint(0, vocab_size, (batch_size, seq_len))

# get token embedding
x = ep[input_ids]
print(x.shape)

# get position embedding
x = x + pe[:, :x.shape[1], :]



print(x.shape)
```
